```python
import numpy as np
from sklearn.datasets import load_diabetes

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
## This will work both in 1 or more dimension, lower and higher dimensions
class MultipleLinearRegression:
    
    def __init__(self):
        self.weights = None
        self.intercept = None
        
    
    def fit(self, X_train, y_train):
        print(f"X_train shape: {X_train.shape}")
        print(f"y_train shape: {y_train.shape}")
        
        
        ## Rn, there is no addition of "1" in X_train in the starting column
        ## Change in X_train, changes at what index, value, which axis
        X_train = np.insert(X_train, 0 , 1, axis=1)
        print(f"New X_train shape: {X_train.shape}")

        ## Calculate Coefficients
        coefficients = np.dot(np.linalg.inv(np.dot(X_train.T, X_train)), X_train.T).dot(y_train)
        print(f"coefficients: {coefficients}")
        print(f"coefficients shape: {coefficients.shape}")
        
        self.weights = coefficients[1:]
        self.intercept = coefficients[0]
        
    
    def predict(self, X_test):
        print(f"X_test shape: {X_test.shape}")
        
        y_pred = np.dot(X_test, self.weights) + self.intercept
        
        return y_pred
```

```python
mlr = MultipleLinearRegression()

mlr.fit(X_train, y_train)
```

```python
y_pred = mlr.predict(X_test)
print(y_pred)
```

```python
r2_score(y_test, y_pred)
```
